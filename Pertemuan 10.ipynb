1. Sorting Arrays

1.1 Introduction to Array Sorting
Sorting adalah proses mengurutkan elemen array, misal dari nilai terkecil ke terbesar. Fungsi np.sort() menghasilkan array baru yang sudah terurut sehingga array asli tidak berubah.

```python
# 1.2 Sorting a One-Dimensional Array (1D)

scores = np.array([88, 72, 95, 80, 76])

print("Original:", scores)
print("Ascending:", np.sort(scores))
print("Descending:", np.sort(scores)[::-1])
print("Original remains:", scores)
```

```python
# 1.3 Sorting a Two-Dimensional Array (2D)

measurement_matrix = np.array([[8, 3, 6],
                               [9, 2, 5],
                               [4, 7, 1]])

print("Original matrix:")
print(measurement_matrix)
print("Sorted within each row:")
print(np.sort(measurement_matrix))
```

```python
# 1.4 Sorting by Axis

print("Sorted by columns (axis=0):")
print(np.sort(measurement_matrix, axis=0))

print("Sorted by rows (axis=1):")
print(np.sort(measurement_matrix, axis=1))

print("All elements sorted:", np.sort(measurement_matrix, axis=None))
```

```python
# 1.5 Sorting Indices with np.argsort()

students = np.array(["Ayu", "Budi", "Citra", "Doni"])
final_scores = np.array([81, 94, 76, 88])
order = np.argsort(final_scores)

print("Sorted scores:", final_scores[order])
print("Corresponding students:", students[order])
```

2. Filtering Arrays

```python
# 2.1 Understanding Boolean Masks

exam_scores = np.array([65, 78, 84, 92, 71, 88])
mask = exam_scores >= 80

print("Scores:", exam_scores)
print("Mask >= 80:", mask)
```

```python
# 2.2 Boolean Indexing

print("Scores >= 80:", exam_scores[exam_scores >= 80])
print("Scores < 80:", exam_scores[exam_scores < 80])
```

```python
# 2.3 Boolean Indexing in a Two-Dimensional Array

student_marks = np.array([[75, 82, 80],
                          [90, 93, 88],
                          [68, 74, 79],
                          [85, 87, 91]])

selected_rows = student_marks[student_marks[:, 0] >= 80]
print("Students with first score >= 80:")
print(selected_rows)
print("Selected shape:", selected_rows.shape)
```

```python
student_marks[:, 0]
```

```python
# urutannya adalah slicing dari semua baris (:) kolom pertama (0) yang first score lebih dari 80
# lalu logika boleannya untuk menyaring memilih baris baris sesuai kondisi booelan dengan semua kolom terpilih [kondisi,]
student_marks[student_marks[:, 0] >= 80]
```

```python
# 2.4 Fancy Indexing

values = np.array([10, 20, 30, 40, 50])
print("Selected positions:", values[[0, 2, 4]])

print("Selected rows:")
print(student_marks[[0, 3], :])
```

```python
# 2.5 Filtering with np.where()

rainfall = np.array([0, 12, 8, 25, 4, 18])
print("Indices with rainfall >= 10:", np.where(rainfall >= 10)[0])
print("Values with rainfall >= 10:", rainfall[np.where(rainfall >= 10)])

print("Rain / No rain:", np.where(rainfall > 0, "Rain", "No rain"))
```

```python
np.where(rainfall >= 10)
```

```python
# 2.6 Conditional Selection with Multiple Conditions

temperatures = np.array([26, 29, 32, 35, 28, 31])

print("28 to 32:", temperatures[(temperatures >= 28) & (temperatures <= 32)])
print("Below 28 or above 32:", temperatures[(temperatures < 28) | (temperatures > 32)])
print("Not above 30:", temperatures[~(temperatures > 30)])
```

3. Updating Arrays

```python
# 3.1 Updating Individual Array Elements

original_data = np.array([10, 20, 30, 40])
updated_data = original_data.copy()
updated_data[1] = 25

print("Original:", original_data)
print("Updated:", updated_data)
```

```python
# 3.2 Updating Rows and Columns

marks = np.array([[70, 75, 80],
                  [82, 86, 88],
                  [90, 91, 93]])

marks[0, 0] = 78
marks[1, :] = [84, 87, 90]
marks[:, 2] = 95

print("Updated marks:")
print(marks)
```

```python
#3.3 Conditional Replacement

measurements = np.array([12, -2, 18, -5, 9])
cleaned = measurements.copy()
cleaned[cleaned < 0] = 0

print("Original:", measurements)
print("Updated with mask:", cleaned)
print("Updated with np.where:", np.where(measurements < 0, 0, measurements))
```

```python
# 3.4 Adding Rows with np.concatenate()

sales = np.array([[12, 15],
                  [10, 18]])
new_row = np.array([[14, 16]])

sales_with_row = np.concatenate((sales, new_row), axis=0)
print("Added row:")
print(sales_with_row)
print("New shape:", sales_with_row.shape)
```

```python
# 3.5 Adding Columns with np.concatenate() 

new_column = np.array([[20], [21], [22]])
sales_with_column = np.concatenate((sales_with_row, new_column), axis=1)

print("Added column:")
print(sales_with_column)
print("New shape:", sales_with_column.shape)
```

```python
# 3.6 Deleting Rows and Columns with np.delete()

print("Delete second row:")
print(np.delete(sales_with_column, 1, axis=0))
print("Delete third column:")
print(np.delete(sales_with_column, 2, axis=1))
```

```python
# 3.7 Shape and Dimension 

matrix = np.arange(1, 7).reshape(2, 3)
row_values = np.array([7, 8, 9]).reshape(1, 3)
column_values = np.array([10, 11]).reshape(2, 1)

print("Matrix shape:", matrix.shape)
print("New row shape:", row_values.shape)
print("New column shape:", column_values.shape)

print("With row:")
print(np.concatenate((matrix, row_values), axis=0))

print("With column:")
print(np.concatenate((matrix, column_values), axis=1))
```

4. Array Aggregation

```python
# 4.1 Sum, Minimum, Maximum, and Mean

daily_visitors = np.array([120, 135, 128, 142, 150])

print("Total:", np.sum(daily_visitors))
print("Minimum:", np.min(daily_visitors))
print("Maximum:", np.max(daily_visitors))
print("Mean:", np.mean(daily_visitors))
```

```python
# 4.2 Aggregation by Axis

test_scores = np.array([[80, 85, 90],
                        [70, 75, 80],
                        [90, 92, 94]])

print("Total per test (axis=0):", np.sum(test_scores, axis=0))
print("Mean per student (axis=1):", np.mean(test_scores, axis=1))
print("Maximum per test (axis=0):", np.max(test_scores, axis=0))
```

```python
# 4.3 Cumulative 
weekly_sales = np.array([10, 15, 12, 18])

print("Weekly sales:", weekly_sales)
print("Cumulative sales:", np.cumsum(weekly_sales))

print("Cumulative scores across columns:")
print(np.cumsum(test_scores, axis=1))
```

```python
# 4.4 The keepdims Argument

row_means = np.mean(test_scores, axis=1)
row_means_2d = np.mean(test_scores, axis=1, keepdims=True)

print("Without keepdims:", row_means, "shape:", row_means.shape)
print("With keepdims:")
print(row_means_2d)
print("Shape:", row_means_2d.shape)
```

```python
# 4.5 Applying Aggregation to Statistical 

rainfall_matrix = np.array([[12, 8, 5],
                            [10, 15, 7],
                            [20, 11, 9],
                            [8, 6, 4]])

print("Total per location:", np.sum(rainfall_matrix, axis=0))
print("Mean per day:", np.mean(rainfall_matrix, axis=1))
print("Overall maximum:", np.max(rainfall_matrix))
```

5. Vectorized Operations

```python
# 5.1 Vectorization vs. Python Loops

values = np.array([2, 4, 6, 8])

# Python loop
loop_result = []
for value in values:
    loop_result.append(value * 2)

# NumPy vectorization
vector_result = values * 2

print("Loop:", loop_result)
print("Vectorized:", vector_result)
```

```python
# 5.2 Scalar Arithmetic Operations

data = np.array([10, 20, 30])

print("Add 5:", data + 5)
print("Subtract 2:", data - 2)
print("Multiply by 3:", data * 3)
print("Divide by 2:", data / 2)
print("Squared:", data ** 2)
```

```python
# 5.3 Element-wise Array Mathematics

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Powers:", a ** 2)
```

```python
# 5.4 Element-wise Comparison Operations

actual = np.array([75, 90, 82, 60])
target = np.array([80, 85, 80, 70])

print("Actual >= target:", actual >= target)
print("Actual == target:", actual == target)
print("Passed target:", actual[actual >= target])
```

```python
# 5.5 Shape Compatibility and Broadcasting

matrix = np.array([[10, 20, 30],
                   [40, 50, 60]])
adjustment = np.array([1, 2, 3])

print("Matrix shape:", matrix.shape)
print("Adjustment shape:", adjustment.shape)
print("Element-wise result:")
print(matrix + adjustment)
```

```python
# 5.6 Applied Example: Standardizing Daily Measurements

observations = np.array([12.0, 15.0, 13.0, 20.0])
mean_value = np.mean(observations)
deviations = observations - mean_value

print("Mean:", mean_value)
print("Deviations:", deviations)
print("Sum of deviations:", np.sum(deviations))
```

Exercise 1 — Sorting Arrays
1. Buat array 1D temperatures berisi [30, 27, 33, 29, 31, 28].
2. Urutkan temperatures secara menaik dan menurun tanpa mengubah array asli.
3. Buat matriks rainfall berisi [[8, 3, 6], [2, 9, 5], [7, 1, 4]].
4. Urutkan rainfall menggunakan axis=0 dan axis=1, lalu jelaskan perbedaannya.

```python
import numpy as np

# 1. Array suhu
temperatures = np.array([30, 27, 33, 29, 31, 28])

ascending = np.sort(temperatures)
descending = np.sort(temperatures)[::-1]

print("Suhu asli:", temperatures)
print("Urutan menaik:", ascending)
print("Urutan menurun:", descending)

# 2. Matriks curah hujan
rainfall = np.array([
    [8, 3, 6],
    [2, 9, 5],
    [7, 1, 4]
])

print("\nMatriks asli:")
print(rainfall)

print("\nSorting axis=0:")
print(np.sort(rainfall, axis=0))

print("\nSorting axis=1:")
print(np.sort(rainfall, axis=1))
```

axis=0 — mengurutkan setiap kolom dari atas ke bawah.
axis=1 — mengurutkan setiap baris dari kiri ke kanan.

axis=0 bekerja pada arah vertikal setiap kolom, sedangkan axis=1 bekerja pada arah horizontal setiap baris.

Exercise 2 — Filtering Arrays

Gunakan scores = np.array([65, 80, 92, 74, 88, 95, 70]).

1. Tampilkan elemen yang bernilai minimal 80 menggunakan Boolean mask.
2. Tampilkan nilai antara 70 dan 90 (termasuk keduanya).
3. Gunakan Fancy indexing untuk mengambil elemen pada indeks 0, 3, dan 5.
4. Gunakan np.where() untuk menemukan indeks nilai yang lebih besar dari 90.

```python
scores = np.array([65, 80, 92, 74, 88, 95, 70])

# 1. Nilai minimal 80
mask = scores >= 80
print("Mask minimal 80:", mask)
print("Nilai minimal 80:", scores[mask])

# 2. Nilai antara 70 dan 90, termasuk batasnya
selected_scores = scores[(scores >= 70) & (scores <= 90)]
print("Nilai antara 70 dan 90:", selected_scores)

# 3. Fancy indexing pada indeks 0, 3, dan 5
selected_positions = scores[[0, 3, 5]]
print("Indeks 0, 3, 5:", selected_positions)

# 4. Indeks nilai yang lebih besar dari 90
indices = np.where(scores > 90)[0]
print("Indeks nilai > 90:", indices)
```

Exercise 3 — Updating Arrays

Gunakan inventory = np.array([[10, -1], [15, 8], [12, -3]]).

1. Buat salinan dan ganti seluruh nilai negatif menjadi 0.
2. Tambahkan baris baru [[20, 5]] menggunakan np.concatenate().
3. Tambahkan kolom [[1], [2], [3], [4]] pada hasil langkah 2.
4. Hapus baris pertama dan kolom terakhir menggunakan np.delete(); periksa .shape hasil akhir.

```python
inventory = np.array([
    [10, -1],
    [15, 8],
    [12, -3]
])

# 1. Salin array dan ganti nilai negatif menjadi 0
clean_inventory = inventory.copy()
clean_inventory[clean_inventory < 0] = 0

print("Inventory setelah diperbaiki:")
print(clean_inventory)

# 2. Tambahkan baris baru
new_row = np.array([[20, 5]])
inventory_rows = np.concatenate(
    (clean_inventory, new_row), axis=0
)

print("\nSetelah menambahkan baris:")
print(inventory_rows)

# 3. Tambahkan kolom baru
new_column = np.array([[1], [2], [3], [4]])
inventory_updated = np.concatenate(
    (inventory_rows, new_column), axis=1
)

print("\nSetelah menambahkan kolom:")
print(inventory_updated)

# 4. Hapus baris pertama dan kolom terakhir
result = np.delete(inventory_updated, 0, axis=0)
result = np.delete(result, -1, axis=1)

print("\nHasil akhir:")
print(result)
print("Shape:", result.shape)
```

Exercise 4 — Array Aggregation

Gunakan visitors = np.array([[20, 25, 30], [18, 22, 28], [24, 27, 32]]).

1. Hitung total dan rata-rata seluruh pengamatan.
2. Hitung total setiap kolom (axis=0) dan rata-rata setiap baris (axis=1).
3. Hitung jumlah kumulatif pada setiap baris (np.cumsum).
4. Hitung rata-rata setiap baris menggunakan keepdims=True, lalu periksa bentuknya.

```python
visitors = np.array([
    [20, 25, 30],
    [18, 22, 28],
    [24, 27, 32]
])

# 1. Total dan rata-rata seluruh pengamatan
print("Total:", np.sum(visitors))
print("Rata-rata:", np.mean(visitors))

# 2. Total setiap kolom dan rata-rata setiap baris
print("Total setiap kolom:", np.sum(visitors, axis=0))
print("Rata-rata setiap baris:", np.mean(visitors, axis=1))

# 3. Jumlah kumulatif setiap baris
print("Cumulative sum:")
print(np.cumsum(visitors, axis=1))

# 4. Rata-rata setiap baris dengan keepdims
row_means = np.mean(visitors, axis=1, keepdims=True)
print("Rata-rata dengan keepdims:")
print(row_means)
print("Shape:", row_means.shape)
```

Exercise 5 — Vectorized Operations

Gunakan prices = np.array([10000, 15000, 20000, 25000]) dan quantities = np.array([2, 3, 1, 4]).

1. Hitung pendapatan tiap produk dengan perkalian element-wise.
2. Hitung total pendapatan menggunakan np.sum().
3. Berikan diskon 10% pada seluruh harga dengan operasi vektorisasi.
4. Tampilkan Boolean mask untuk produk dengan pendapatan minimal 50000.

```python
prices = np.array([10000, 15000, 20000, 25000])
quantities = np.array([2, 3, 1, 4])

# 1. Pendapatan setiap produk
revenue = prices * quantities
print("Pendapatan setiap produk:", revenue)

# 2. Total pendapatan
total_revenue = np.sum(revenue)
print("Total pendapatan:", total_revenue)

# 3. Diskon 10% pada setiap harga
discounted_prices = prices * 0.9
print("Harga setelah diskon 10%:", discounted_prices)

# 4. Boolean mask pendapatan minimal 50000
mask = revenue >= 50000
print("Mask pendapatan >= 50000:", mask)
```

Exercise 6 — Integrated Statistical Practice

Gunakan data nilai tiga ujian dari empat mahasiswa:

student_results = np.array([[80, 75, 90],
                            [65, 70, 72],
                            [92, 88, 95],
                            [78, 85, 82]])
1. Hitung rata-rata nilai masing-masing mahasiswa.
2. Pilih baris mahasiswa yang rata-ratanya minimal 80.
3. Tentukan posisi mahasiswa dengan nilai rata-rata tertinggi memakai np.argsort().
4. Tambahkan 5 poin pada setiap nilai, tetapi batasi nilai maksimum menjadi 100 dengan np.where().
5. Hitung rata-rata per ujian pada data yang telah diperbarui.

```python
student_results = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [92, 88, 95],
    [78, 85, 82]
])

# 1. Rata-rata nilai setiap mahasiswa
student_means = np.mean(student_results, axis=1)
print("Rata-rata setiap mahasiswa:", student_means)

# 2. Pilih mahasiswa dengan rata-rata minimal 80
selected_students = student_results[student_means >= 80]
print("\nNilai mahasiswa dengan rata-rata >= 80:")
print(selected_students)

# 3. Urutan posisi mahasiswa berdasarkan rata-rata
order = np.argsort(student_means)
print("\nUrutan indeks dari rata-rata terendah:", order)
print("Indeks rata-rata tertinggi:", order[-1])

# 4. Tambahkan 5 poin, maksimum 100
updated_results = np.where(
    student_results + 5 > 100,
    100,
    student_results + 5
)

print("\nNilai setelah penambahan 5 poin:")
print(updated_results)

# 5. Rata-rata setiap ujian setelah diperbarui
exam_means = np.mean(updated_results, axis=0)
print("\nRata-rata setiap ujian:", exam_means)
```
